```python
def sampleACVF(X, h):
    n = len(X)
    x_mean = np.mean(X)
    return 1/n*sum((X[i]-x_mean)*(X[i+abs(h)]-x_mean) for i in range(1,n-abs(h)))
def sampleACF(X, h):
    return sampleACVF(X, h)/sampleACVF(X, 0)

hs = np.arange(40)
acc = [sampleACF(np.array(df_spy['Returns']), h) for h in hs]
plt.vlines(x=hs, ymin=0, ymax=acc, color = 'hotpink')
plt.scatter(hs, acc, marker = '*', color = 'hotpink', label = 'ACF')
plt.title('ACF'); plt.legend()
plt.xlabel('h')
```

```python
data = df_spy['Returns'].values

df = pd.DataFrame()
ps = []
qs = []
BIC = []
AIC = []
HQIC = []

max_p = 5
max_q = 5

for p in range(0, max_p):
    for q in range(0, max_q):
        model = ARIMA(data, order=(p, 0, q))
        model_fit = model.fit()
        ps.append(p)
        qs.append(q)
        AIC.append(model_fit.aic)
        BIC.append(model_fit.bic)
        HQIC.append(model_fit.hqic)

df['p']    = ps
df['q']    = qs
df['AIC']  = AIC
df['BIC']  = BIC
df['HQIC'] = HQIC
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.arima.model import ARIMA
from arch import arch_model
from sklearn.metrics import accuracy_score, classification_report

# 1. data preparation
spy = pd.read_csv(spy_path, index_col='Date', parse_dates=True)
spy.sort_index(ascending=True, inplace=True)
spy['Returns'] = np.log(spy['Close'] / spy['Close'].shift(1))
spy.dropna(inplace=True)

# Window settings
window = 20
pred_returns = []
pred_vols = []

print("Calculating forecasts (Rolling Window 20d)...")

for i in range(len(spy) - window):
    train_window = spy['Returns'].iloc[i:i+window]
    
    # --- MODEL 1: ARIMA (Returns) ---
    # params: (2,0,0) 
    try:
        arima = ARIMA(train_window, order=(2, 0, 0)).fit()
        f_ret = arima.forecast(steps=1).values[0]
    except: f_ret = 0
        
    # --- MODEL 2: GARCH (Volatility) ---
    # params: (1,1) - standard model
    try:
        garch = arch_model(train_window, vol='Garch', p=1, q=1, rescale=False).fit(disp='off')
        f_vol = np.sqrt(garch.forecast(horizon=1).variance.values[-1][0])
    except: f_vol = spy['Returns'].std()

    pred_returns.append(f_ret)
    pred_vols.append(f_vol)

# comparison DataFrame
df_comp = spy.iloc[window:].copy()
df_comp['Pred_Return'] = pred_returns
df_comp['Pred_Vol'] = pred_vols
df_comp['Actual_Dir'] = (df_comp['Returns'] > 0).astype(int)
df_comp['Pred_Dir'] = (df_comp['Pred_Return'] > 0).astype(int)
spy_cum = (1 + (np.exp(df_comp['Returns']) - 1)).cumprod()
# --- COMPARATIVE ANALYSIS ---

fig, axs = plt.subplots(3, 1, figsize=(14, 15))

# 1. Comparison of Returns (Regression)
axs[0].plot(df_comp['Returns'].tail(150), label='Actual Returns', color='black', alpha=0.3)
axs[0].plot(df_comp['Pred_Return'].tail(150), label='ARIMA Prediction', color='blue', linewidth=2)
axs[0].set_title('COMPARISON 1: Returns (Last 150 Days)')
axs[0].legend()

# 2. Comparison of Volatility (Risk)
axs[1].plot(df_comp.index[-150:], df_comp['Returns'].tail(150), color='gray', alpha=0.2, label='Returns')
axs[1].plot(df_comp['Pred_Vol'].tail(150), label='Volatility Prediction (GARCH)', color='red', linewidth=2)
axs[1].set_title('COMPARISON 2: Volatility / Risk')
axs[1].legend()

# 3. Cumulative Return (Portfolio Effectiveness)
spy_cum = (1 + (np.exp(df_comp['Returns']) - 1)).cumprod()
strat_cum = (1 + (df_comp['Pred_Dir'] * (np.exp(df_comp['Returns']) - 1))).cumprod()
axs[2].plot(spy_cum, label='Benchmark: Buy and Hold SPY', color='black', linestyle='--')
axs[2].plot(strat_cum, label='Strategy: Trade Based on Direction Prediction', color='green', linewidth=2)
axs[2].set_title('COMPARISON 3: Capital (ARIMA Direction)')
axs[2].legend()

plt.tight_layout()
plt.show()

# --- STATISTICAL SUMMARY ---
acc = accuracy_score(df_comp['Actual_Dir'], df_comp['Pred_Dir'])
print("-" * 30)
print(f"SUMMARY OF SP500 MODEL:")
print(f"Direction Accuracy: {acc:.2%}")
print("-" * 30)
print("Classification Report:")
print(classification_report(df_comp['Actual_Dir'], df_comp['Pred_Dir'], target_names=['DOWN', 'UP']))
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from arch import arch_model
import warnings

warnings.filterwarnings("ignore")

# 1. Przygotowanie danych bazowych
spy = pd.read_csv(spy_path, index_col='Date', parse_dates=True)
spy.sort_index(ascending=True, inplace=True)
spy['Returns'] = np.log(spy['Close'] / spy['Close'].shift(1))

# 2. Wskaźniki Długoterminowe (Obliczane na całym szeregu - szybkie)
# Średnia krocząca 200 dni (Trend długoterminowy)
spy['MA200_Trend'] = spy['Close'].rolling(window=200).mean()
spy['Dist_From_MA200'] = (spy['Close'] - spy['MA200_Trend']) / spy['MA200_Trend']

# 3. Pętla Krocząca dla ARIMA (20d) i GARCH (60d)
# Zaczynamy od 200, żeby MA200 było już gotowe
start_idx = 200
arima_preds = []
garch_vols = []
dates = []

print("Generuję zaawansowane cechy (ARIMA 20d, GARCH 60d)...")

for i in range(start_idx, len(spy)):
    # --- KRÓTKIE OKNO (20d) dla ARIMA ---
    window_20 = spy['Returns'].iloc[i-20:i]
    try:
        model_arima = ARIMA(window_20, order=(1, 0, 1)).fit()
        pred_ret = model_arima.forecast(steps=1).values[0]
    except: pred_ret = 0
    
    # --- DŁUŻSZE OKNO (60d) dla GARCH ---
    # Zmienność potrzebuje więcej danych, by "zauważyć" trend strachu
    window_60 = spy['Returns'].iloc[i-60:i]
    try:
        model_garch = arch_model(window_60, vol='Garch', p=1, q=1, rescale=False).fit(disp='off')
        pred_vol = np.sqrt(model_garch.forecast(horizon=1).variance.values[-1][0])
    except: pred_vol = spy['Returns'].iloc[:i].std()

    arima_preds.append(pred_ret)
    garch_vols.append(pred_vol)
    dates.append(spy.index[i])

# 4. Łączenie wszystkiego w jedną tabelę (Feature Matrix)
df_features = spy.loc[dates].copy()
df_features['ARIMA_20d_Pred'] = arima_preds
df_features['GARCH_60d_Vol'] = garch_vols

# Czyścimy NaN (z MA200)
df_features.dropna(inplace=True)

# 5. Podsumowanie i podgląd
print("\nGotowy Feature Matrix dla modeli ML:")
print(df_features[['Close', 'Dist_From_MA200', 'ARIMA_20d_Pred', 'GARCH_60d_Vol']].tail())

# Wizualizacja dla raportu: Porównanie zmienności 20d vs 60d (opcjonalnie)
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 5))
plt.plot(df_features['GARCH_60d_Vol'], color='purple', label='GARCH Volatility (60d Window)')
plt.title('Stabilny pomiar ryzyka (GARCH 60d)')
plt.legend()
plt.show()
```

```python
import os
import pandas as pd
import numpy as np

# Paths

# 1. SECTOR ANALYSIS (Ranking)
sector_data = []
for ticker in set(sector_map.values()):
    file_path = os.path.join(etf_path, f"{ticker.lower()}_clean.csv")
    if os.path.exists(file_path):
        df_etf = pd.read_csv(file_path, index_col='Date', parse_dates=True)
        df_etf.sort_index(ascending=True, inplace=True)
        # Obliczamy momentum (zwrot 20d) i zmienność (GARCH-like std)
        returns = np.log(df_etf['Close'] / df_etf['Close'].shift(1))
        score = returns.tail(20).mean() / returns.tail(20).std() # Sharpe-like score
        
        sector_data.append({
            'Sector': ticker,
            'Sector_Score': score,
            'Sector_Trend': 'UP' if df_etf['Close'].iloc[-1] > df_etf['Close'].rolling(200).mean().iloc[-1] else 'DOWN'
        })

df_sector_rank = pd.DataFrame(sector_data).sort_values(by='Sector_Score', ascending=False)

# 2. INTEGRACJA Z TWOIM RANKINGIEM AKCJI (ranking_df z poprzedniego kroku)
# Dodajemy kolumnę sektora do Twoich wyników LightGBM
ranking_df['Sector'] = ranking_df['Ticker'].map(sector_map)

# Łączymy wyniki akcji z wynikami ich sektorów
final_decision_df = ranking_df.merge(df_sector_rank, on='Sector', how='left')

# 3. FINALNY FILTR (The "Gold Standard" Selection)
# Kryteria: 
# - Wysokie prawdopodobieństwo wzrostu z LightGBM (Prob_Growth)
# - Sektor musi być w trendzie wzrostowym (Sector_Trend == UP)
# - Sektor musi mieć dodatni wynik (Sector_Score > 0)

top_picks = final_decision_df[
    (final_decision_df['Sector_Trend'] == 'UP') & 
    (final_decision_df['Sector_Score'] > 0)
].sort_values(by='Prob_Growth', ascending=False)

# --- WIZUALIZACJA WYNIKÓW ---
print("\n" + "="*60)
print("             RAPORT DECYZYJNY: SEKTOR + AKCJA")
print("="*60)
if not top_picks.empty:
    print(top_picks[['Ticker', 'Sector', 'Prob_Growth', 'Sector_Score', 'Trend']].head(5).to_string(index=False))
else:
    print("BRAK SILNYCH SYGNAŁÓW: Rynek w trendzie spadkowym lub wysokie ryzyko.")
print("="*60)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
from statsmodels.tsa.arima.model import ARIMA
from arch import arch_model
import warnings

warnings.filterwarnings("ignore")

# --- KONFIGURACJA ŚCIEŻEK ---
target_date = '2015-03-01'  # <--- TU WPISZ DATĘ TESTU
etf_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/ETF clean'
stocks_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/Stocks clean'
spy_path = os.path.join(etf_path, 'spy_clean.csv')

# --- MAPOWANIE SEKTORÓW ---
sector_map = {
    'AAPL': 'XLK', 'MSFT': 'XLK', 'NVDA': 'XLK', 'AVGO': 'XLK', 'ORCL': 'XLK', 'ADBE': 'XLK',
    'LLY': 'XLV', 'JNJ': 'XLV', 'UNH': 'XLV', 'ABBV': 'XLV', 'MRK': 'XLV',
    'JPM': 'XLF', 'BAC': 'XLF', 'WFC': 'XLF', 'GS': 'XLF', 'MS': 'XLF',
    'GOOGL': 'XLC', 'META': 'XLC', 'NFLX': 'XLC', 'DIS': 'XLC',
    'AMZN': 'XLY', 'TSLA': 'XLY', 'HD': 'XLY', 'NKE': 'XLY',
    'PG': 'XLP', 'KO': 'XLP', 'PEP': 'XLP'
}

def run_trading_simulation(date_input, days_horizon=22):
    print(f"--- URUCHAMIAM ANALIZĘ NA DZIEŃ: {date_input} ---")
    
    # 1. ANALIZA SEKTORÓW (Ranking "Wczoraj")
    sector_data = []
    for ticker in set(sector_map.values()):
        file_path = os.path.join(etf_path, f"{ticker.lower()}_clean.csv")
        if os.path.exists(file_path):
            df = pd.read_csv(file_path, index_col='Date', parse_dates=True).sort_index()
            df_cut = df.loc[:date_input]
            
            if len(df_cut) < 200: continue
            
            rets = np.log(df_cut['Close'] / df_cut['Close'].shift(1))
            score = rets.tail(20).mean() / rets.tail(20).std()
            ma200 = df_cut['Close'].rolling(200).mean().iloc[-1]
            sector_data.append({
                'Sector': ticker,
                'Sector_Score': score,
                'Sector_Trend': 'UP' if df_cut['Close'].iloc[-1] > ma200 else 'DOWN'
            })
    
    df_sector_rank = pd.DataFrame(sector_data)
    
    # 2. SKANOWANIE AKCJI (Micro-Analysis)
    stock_results = []
    for ticker, sector in sector_map.items():
        file_path = os.path.join(stocks_path, f"{ticker.lower()}_clean.csv")
        if os.path.exists(file_path):
            df = pd.read_csv(file_path, index_col='Date', parse_dates=True).sort_index()
            df_cut = df.loc[:date_input]
            if len(df_cut) < 60: continue
            
            # --- FEATURE ENGINEERING (Dla LightGBM) ---
            rets = np.log(df_cut['Close'] / df_cut['Close'].shift(1)).dropna()
            
            # ARIMA 20d
            try:
                m_arima = ARIMA(rets.tail(20), order=(1,0,1)).fit()
                pred_arima = m_arima.forecast(1).values[0]
            except: pred_arima = 0
            
            # GARCH 60d
            try:
                m_garch = arch_model(rets.tail(60), vol='Garch', p=1, q=1, rescale=False).fit(disp='off')
                pred_vol = np.sqrt(m_garch.forecast(horizon=1).variance.values[-1][0])
            except: pred_vol = rets.std()
            
            # Tutaj symulujemy wynik modelu (w realnym kodzie: model.predict())
            # Im wyższa ARIMA i niższy GARCH, tym wyższy scoring Prob_Growth
            prob = 0.5 + (pred_arima * 10) - (pred_vol * 0.1) 
            
            stock_results.append({
                'Ticker': ticker, 'Sector': sector, 'Prob_Growth': prob
            })
            
    df_stocks = pd.DataFrame(stock_results)
    
    # 3. FILTROWANIE I WYBÓR TOP 3
    final_df = df_stocks.merge(df_sector_rank, on='Sector')
    top_picks = final_df[
        (final_df['Sector_Trend'] == 'UP') & (final_df['Sector_Score'] > 0)
    ].sort_values(by='Prob_Growth', ascending=False).head(3)
    
    if top_picks.empty:
        print("Brak bezpiecznych sygnałów. Rynek w trendzie spadkowym.")
        return
    
    # 4. BACKTEST (Symulacja przyszłości)
    tickers = top_picks['Ticker'].tolist()
    perf_df = pd.DataFrame()
    
    # Benchmark SPY
    spy = pd.read_csv(spy_path, index_col='Date', parse_dates=True).sort_index()
    spy_after = spy.loc[date_input:].iloc[:days_horizon]
    perf_df['SP500'] = (spy_after['Close'] / spy_after['Close'].iloc[0]) * 100
    
    # Portfel
    for t in tickers:
        s_df = pd.read_csv(os.path.join(stocks_path, f"{t.lower()}_clean.csv"), index_col='Date', parse_dates=True).sort_index()
        s_after = s_df.loc[date_input:].iloc[:days_horizon]
        perf_df[t] = (s_after['Close'] / s_after['Close'].iloc[0]) * 100
    
    perf_df['PORTFOLIO'] = perf_df[tickers].mean(axis=1)
    
    # 5. WIZUALIZACJA
    plt.figure(figsize=(12, 6))
    for t in tickers:
        plt.plot(perf_df.index, perf_df[t], alpha=0.3, label=t)
    
    plt.plot(perf_df.index, perf_df['PORTFOLIO'], color='green', lw=3, label='MODEL PORTFOLIO')
    plt.plot(perf_df.index, perf_df['SP500'], color='black', ls='--', lw=2, label='SP500 BENCHMARK')
    
    plt.title(f'Test for date {target_date} ({days_to_track} days)')
    plt.ylabel('Value (Start = 100)')
    plt.legend()
    plt.grid(True, alpha=0.2)
    plt.show()
    
    # Finalne statystyki
    ret_model = perf_df['PORTFOLIO'].iloc[-1] - 100
    ret_spy = perf_df['SP500'].iloc[-1] - 100
    print(f"chosen stocks: {tickers}")
    print(f"Portfolio Return: {ret_model:.2f}%")
    print(f"SP500 Return:    {ret_spy:.2f}%")
    print(f"Alpha: {ret_model - ret_spy:.2f} pp.")

# Uruchomienie
run_trading_simulation(target_date)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
from statsmodels.tsa.arima.model import ARIMA
from arch import arch_model
import warnings

warnings.filterwarnings("ignore")

# --- KONFIGURACJA ŚCIEŻEK ---
# target_date = '2023-01-05' 
etf_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/ETF clean'
stocks_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/Stocks clean'
spy_path = os.path.join(etf_path, 'spy_clean.csv')

sector_map = {
    'AAPL': 'XLK', 'MSFT': 'XLK', 'NVDA': 'XLK', 'AVGO': 'XLK', 'ORCL': 'XLK', 'ADBE': 'XLK',
    'LLY': 'XLV', 'JNJ': 'XLV', 'UNH': 'XLV', 'ABBV': 'XLV', 'MRK': 'XLV',
    'JPM': 'XLF', 'BAC': 'XLF', 'WFC': 'XLF', 'GS': 'XLF', 'MS': 'XLF',
    'GOOGL': 'XLC', 'META': 'XLC', 'NFLX': 'XLC', 'DIS': 'XLC',
    'AMZN': 'XLY', 'TSLA': 'XLY', 'HD': 'XLY', 'NKE': 'XLY',
    'PG': 'XLP', 'KO': 'XLP', 'PEP': 'XLP'
}

# =================================================================
# KROK 1: TRENING MODELU (Budujemy "mózg" na danych historycznych)
# =================================================================
def train_lightgbm_model():
    print("Rozpoczynam trening modelu LightGBM na danych SPY...")
    spy_train = pd.read_csv(spy_path, index_col='Date', parse_dates=True).sort_index()
    # Trenujemy tylko na danych przed target_date, żeby nie oszukiwać!
    spy_train = spy_train.loc[:target_date]
    
    spy_train['Returns'] = np.log(spy_train['Close'] / spy_train['Close'].shift(1))
    spy_train['Target'] = (spy_train['Returns'].shift(-1) > 0).astype(int) # Czy jutro rośnie?
    
    # Feature Engineering dla treningu
    spy_train['MA200'] = spy_train['Close'].rolling(200).mean()
    spy_train['Dist_MA200'] = (spy_train['Close'] - spy_train['MA200']) / spy_train['MA200']
    spy_train['Vol_Change'] = spy_train['Volume'].pct_change()
    
    # Uproszczone cechy dla treningu (ARIMA/GARCH dla SPY)
    # W realnym treningu robimy to w pętli, tutaj bierzemy lagowane zwroty jako proxy
    spy_train['Lag_Ret'] = spy_train['Returns'].shift(1)
    
    features = ['Dist_MA200', 'Vol_Change', 'Lag_Ret']
    df_train = spy_train.dropna()
    
    X = df_train[features]
    y = df_train['Target']
    
    train_data = lgb.Dataset(X, label=y)
    params = {'objective': 'binary', 'metric': 'binary_logloss', 'verbose': -1, 'learning_rate': 0.05}
    model = lgb.train(params, train_data, num_boost_round=100)
    return model, features

# Uruchamiamy trening
trained_model, feature_names = train_lightgbm_model()

# =================================================================
# KROK 2: SYMULACJA Z UŻYCIEM WYTRENOWANEGO LIGHTGBM
# =================================================================
def run_trading_simulation(date_input, days_horizon=22):
    print(f"\n--- URUCHAMIAM ANALIZĘ Z LIGHTGBM NA DZIEŃ: {date_input} ---")
    
    # 1. ANALIZA SEKTORÓW
    sector_data = []
    for ticker in set(sector_map.values()):
        file_path = os.path.join(etf_path, f"{ticker.lower()}_clean.csv")
        if os.path.exists(file_path):
            df = pd.read_csv(file_path, index_col='Date', parse_dates=True).sort_index()
            df_cut = df.loc[:date_input]
            if len(df_cut) < 200: continue
            
            rets = np.log(df_cut['Close'] / df_cut['Close'].shift(1))
            score = rets.tail(20).mean() / rets.tail(20).std()
            ma200 = df_cut['Close'].rolling(200).mean().iloc[-1]
            sector_data.append({'Sector': ticker, 'Sector_Score': score, 
                                'Sector_Trend': 'UP' if df_cut['Close'].iloc[-1] > ma200 else 'DOWN'})
    
    df_sector_rank = pd.DataFrame(sector_data)

    # 2. SKANOWANIE AKCJI (PREDYKCJA LIGHTGBM)
    stock_results = []
    for ticker, sector in sector_map.items():
        file_path = os.path.join(stocks_path, f"{ticker.lower()}_clean.csv")
        if os.path.exists(file_path):
            df = pd.read_csv(file_path, index_col='Date', parse_dates=True).sort_index()
            df_cut = df.loc[:date_input]
            if len(df_cut) < 200: continue
            
            # Przygotowanie cech do predykcji
            ma200_val = df_cut['Close'].rolling(200).mean().iloc[-1]
            dist_ma200 = (df_cut['Close'].iloc[-1] - ma200_val) / ma200_val
            vol_chg = df_cut['Volume'].pct_change().iloc[-1]
            lag_ret = np.log(df_cut['Close'].iloc[-1] / df_cut['Close'].iloc[-2])
            
            # Wektor cech dla LightGBM
            X_live = pd.DataFrame([[dist_ma200, vol_chg, lag_ret]], columns=feature_names)
            
            # PREDYKCJA FAKTYCZNYM MODELEM
            prob = trained_model.predict(X_live)[0]
            
            stock_results.append({'Ticker': ticker, 'Sector': sector, 'Prob_Growth': prob})
            
    df_stocks = pd.DataFrame(stock_results)
    final_df = df_stocks.merge(df_sector_rank, on='Sector')
    
    # Wybór Top 3
    top_picks = final_df[
        (final_df['Sector_Trend'] == 'UP') & (final_df['Sector_Score'] > 0)
    ].sort_values(by='Prob_Growth', ascending=False).head(3)
    
    if top_picks.empty:
        print("Brak sygnałów spełniających kryteria.")
        return

    # 3. BACKTEST I PLOT (Bez zmian w logice wizualizacji)
    tickers = top_picks['Ticker'].tolist()
    spy = pd.read_csv(spy_path, index_col='Date', parse_dates=True).sort_index()
    spy_after = spy.loc[date_input:].iloc[:days_horizon]
    
    perf_df = pd.DataFrame()
    perf_df['SP500'] = (spy_after['Close'] / spy_after['Close'].iloc[0]) * 100
    
    for t in tickers:
        s_df = pd.read_csv(os.path.join(stocks_path, f"{t.lower()}_clean.csv"), index_col='Date', parse_dates=True).sort_index()
        s_after = s_df.loc[date_input:].iloc[:days_horizon]
        perf_df[t] = (s_after['Close'] / s_after['Close'].iloc[0]) * 100
    
    perf_df['PORTFOLIO'] = perf_df[tickers].mean(axis=1)
    
    # Plotowanie
    plt.figure(figsize=(12, 6))
    plt.plot(perf_df['PORTFOLIO'], color='green', lw=3, label=f'Model Portfolio {tickers}')
    plt.plot(perf_df['SP500'], color='black', ls='--', label='SP500')
    plt.title(f'LightGBM Strategy Test: {date_input}')
    plt.legend()
    plt.show()
    
    print(f"Top Picks: {tickers}")
    print(f"Portfolio Return: {perf_df['PORTFOLIO'].iloc[-1]-100:.2f}%")
    print(f"SP500 Return: {perf_df['SP500'].iloc[-1]-100:.2f}%")

# Wywołanie
run_trading_simulation(target_date)
```

## Finalna wersja: wybór sektorów i spółek co około 20 dni

Ta część robi symulację bliższą temu, jak model miałby działać w praktyce:

- na dany dzień model zna tylko historię do tej daty,
- ocenia sektory na podstawie trendu i 20-dniowego momentum/skorygowanego o zmienność wyniku,
- dla każdej spółki liczy cechy i predykcję `Prob_Growth`,
- wybiera Top N spółek tylko z sektorów spełniających filtr,
- trzyma wybór przez około 20 dni kalendarzowych, potem wybiera sektor i spółki od nowa,
- rysuje portfel od wybranej daty przez około 1-2 miesiące.


```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import lightgbm as lgb

warnings.filterwarnings("ignore")

# ================================================================
# FINALNA KONFIGURACJA
# ================================================================
etf_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/ETF clean'
stocks_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/Stocks clean'
spy_path = os.path.join(etf_path, 'spy_clean.csv')

# Możesz zmienić te parametry.
target_date = '2015-03-01'
forecast_calendar_days = 60       # 1-2 miesiące kalendarzowe
rebalance_calendar_days = 20      # sektor i spółki wybierane od nowa co ok. 20 dni kalendarzowych
top_n = 3
retrain_model_each_rebalance = True  # walk-forward: przy każdej rotacji model uczy się na nowszych znanych danych

sector_map = {
    'AAPL': 'XLK', 'MSFT': 'XLK', 'NVDA': 'XLK', 'AVGO': 'XLK', 'ORCL': 'XLK', 'ADBE': 'XLK',
    'LLY': 'XLV', 'JNJ': 'XLV', 'UNH': 'XLV', 'ABBV': 'XLV', 'MRK': 'XLV',
    'JPM': 'XLF', 'BAC': 'XLF', 'WFC': 'XLF', 'GS': 'XLF', 'MS': 'XLF',
    'GOOGL': 'XLC', 'META': 'XLC', 'NFLX': 'XLC', 'DIS': 'XLC',
    'AMZN': 'XLY', 'TSLA': 'XLY', 'HD': 'XLY', 'NKE': 'XLY',
    'PG': 'XLP', 'KO': 'XLP', 'PEP': 'XLP'
}

final_features = [
    'Dist_From_MA200',
    'Vol_Change',
    'Intraday_Range',
    'Returns',
    'Momentum_20d',
    'Volatility_20d'
]


def load_price_data(path):
    df = pd.read_csv(path, index_col='Date', parse_dates=True).sort_index()
    df = df[~df.index.duplicated(keep='last')]
    return df


def add_final_features(df):
    df = df.copy()
    df['Returns'] = np.log(df['Close'] / df['Close'].shift(1))
    df['Target'] = (df['Returns'].shift(-1) > 0).astype(int)
    df['MA200'] = df['Close'].rolling(200).mean()
    df['Dist_From_MA200'] = (df['Close'] - df['MA200']) / df['MA200']
    df['Vol_Change'] = df['Volume'].pct_change()
    df['Intraday_Range'] = (df['High'] - df['Low']) / df['Open']
    df['Momentum_20d'] = np.log(df['Close'] / df['Close'].shift(20))
    df['Volatility_20d'] = df['Returns'].rolling(20).std()
    return df


def nearest_trading_day(index, date_like):
    date_like = pd.Timestamp(date_like)
    valid = index[index >= date_like]
    if len(valid) == 0:
        raise ValueError(f'Brak danych giełdowych po dacie {date_like.date()}')
    return valid[0]


def train_final_spy_model(as_of_date):
    spy = add_final_features(load_price_data(spy_path))
    as_of_date = nearest_trading_day(spy.index, as_of_date)

    # Bez look-ahead bias:
    # wiersz z dnia T ma Target = kierunek z T+1, więc w dniu T ta etykieta nie jest jeszcze znana.
    # Przy decyzji po zamknięciu T możemy trenować maksymalnie na wierszach do T-1,
    # bo ich targety są już znane po zamknięciu T.
    as_of_pos = spy.index.get_loc(as_of_date)
    if as_of_pos == 0:
        raise ValueError('Za mało danych przed as_of_date.')
    train_end_date = spy.index[as_of_pos - 1]

    train = spy.loc[:train_end_date].dropna(subset=final_features + ['Target'])
    if len(train) < 300:
        raise ValueError('Za mało danych historycznych do treningu modelu przed as_of_date.')

    X_train = train[final_features]
    y_train = train['Target']

    params = {
        'objective': 'binary',
        'metric': 'binary_logloss',
        'boosting_type': 'gbdt',
        'learning_rate': 0.03,
        'num_leaves': 15,
        'max_depth': 4,
        'feature_fraction': 0.85,
        'bagging_fraction': 0.85,
        'bagging_freq': 5,
        'seed': 42,
        'verbose': -1
    }
    dataset = lgb.Dataset(X_train, label=y_train)
    model = lgb.train(params, dataset, num_boost_round=250)
    return model, X_train, y_train, train_end_date


def score_sectors(as_of_date):
    rows = []
    for sector in sorted(set(sector_map.values())):
        path = os.path.join(etf_path, f'{sector.lower()}_clean.csv')
        if not os.path.exists(path):
            continue

        df = load_price_data(path).loc[:as_of_date]
        if len(df) < 220:
            continue

        returns = np.log(df['Close'] / df['Close'].shift(1)).dropna()
        recent = returns.tail(20)
        score = recent.mean() / recent.std() if recent.std() != 0 else 0
        ma200 = df['Close'].rolling(200).mean().iloc[-1]
        trend = 'UP' if df['Close'].iloc[-1] > ma200 else 'DOWN'

        rows.append({
            'Sector': sector,
            'Sector_Score': score,
            'Sector_Trend': trend
        })

    return pd.DataFrame(rows)


def score_stocks(as_of_date, model):
    rows = []
    feature_rows = []

    for ticker, sector in sector_map.items():
        path = os.path.join(stocks_path, f'{ticker.lower()}_clean.csv')
        if not os.path.exists(path):
            continue

        df = add_final_features(load_price_data(path)).loc[:as_of_date]
        usable = df.dropna(subset=final_features)
        if usable.empty:
            continue

        x_live = usable[final_features].tail(1)
        prob_growth = float(model.predict(x_live)[0])
        feature_row = x_live.copy()
        feature_row.index = [ticker]
        feature_rows.append(feature_row)

        rows.append({
            'Ticker': ticker,
            'Sector': sector,
            'Prob_Growth': prob_growth,
            'Close_Price': float(usable['Close'].iloc[-1]),
            'Trend': 'Bullish' if usable['Dist_From_MA200'].iloc[-1] > 0 else 'Bearish'
        })

    feature_snapshot = pd.concat(feature_rows) if feature_rows else pd.DataFrame(columns=final_features)
    return pd.DataFrame(rows), feature_snapshot


def select_portfolio(as_of_date, model, top_n=3):
    sector_scores = score_sectors(as_of_date)
    stock_scores, feature_snapshot = score_stocks(as_of_date, model)

    if sector_scores.empty or stock_scores.empty:
        return pd.DataFrame(), sector_scores, stock_scores, feature_snapshot

    decision_table = stock_scores.merge(sector_scores, on='Sector', how='left')
    decision_table['Final_Score'] = decision_table['Prob_Growth'] + 0.15 * decision_table['Sector_Score'].fillna(0)

    picks = (
        decision_table[
            (decision_table['Sector_Trend'] == 'UP') &
            (decision_table['Sector_Score'] > 0)
        ]
        .sort_values(['Final_Score', 'Prob_Growth'], ascending=False)
        .head(top_n)
    )

    return picks, sector_scores, decision_table, feature_snapshot


def build_rebalance_dates(trading_index, start_date, end_date, every_calendar_days=20):
    dates = []
    current = pd.Timestamp(start_date)
    end_date = pd.Timestamp(end_date)

    while current <= end_date:
        rebalance_day = nearest_trading_day(trading_index, current)
        if rebalance_day <= end_date and (not dates or rebalance_day != dates[-1]):
            dates.append(rebalance_day)
        current = current + pd.Timedelta(days=every_calendar_days)

    return dates


def run_final_rotating_strategy(
    start_date=target_date,
    forecast_days=forecast_calendar_days,
    rebalance_days=rebalance_calendar_days,
    top_n=top_n,
    retrain_each_rebalance=retrain_model_each_rebalance
):
    spy = load_price_data(spy_path)
    start_trading_day = nearest_trading_day(spy.index, start_date)
    end_calendar_day = start_trading_day + pd.Timedelta(days=forecast_days)
    end_trading_day = spy.index[spy.index <= end_calendar_day][-1]

    # Model może być trenowany raz na start albo w trybie walk-forward przy każdej rotacji.
    # Domyślnie używamy walk-forward, bo wtedy każda decyzja korzysta z najnowszych znanych danych.
    model = None
    X_train_final = None
    y_train_final = None
    train_end_final = None
    if not retrain_each_rebalance:
        model, X_train_final, y_train_final, train_end_final = train_final_spy_model(start_trading_day)

    rebalance_dates = build_rebalance_dates(spy.index, start_trading_day, end_trading_day, rebalance_days)

    portfolio_values = pd.Series(dtype=float)
    portfolio_value = 100.0
    decisions = []
    last_feature_snapshot = pd.DataFrame(columns=final_features)
    last_picks = pd.DataFrame()

    for i, rebalance_day in enumerate(rebalance_dates):
        next_rebalance = rebalance_dates[i + 1] if i + 1 < len(rebalance_dates) else end_trading_day
        interval_dates = spy.loc[rebalance_day:next_rebalance].index
        if len(interval_dates) == 0:
            continue

        if retrain_each_rebalance:
            model, X_train_final, y_train_final, train_end_final = train_final_spy_model(rebalance_day)

        picks, sector_scores, decision_table, feature_snapshot = select_portfolio(rebalance_day, model, top_n=top_n)
        last_feature_snapshot = feature_snapshot
        last_picks = picks

        if picks.empty:
            decisions.append({
                'Rebalance_Date': rebalance_day,
                'Selected_Tickers': 'CASH',
                'Selected_Sectors': 'CASH',
                'Mean_Prob_Growth': np.nan,
                'Model_Trained_Through': train_end_final,
                'Note': 'Brak spółek po filtrze sektorowym'
            })
            for day in interval_dates:
                portfolio_values.loc[day] = portfolio_value
            continue

        tickers = picks['Ticker'].tolist()
        decisions.append({
            'Rebalance_Date': rebalance_day,
            'Selected_Tickers': ', '.join(tickers),
            'Selected_Sectors': ', '.join(sorted(picks['Sector'].unique())),
            'Mean_Prob_Growth': picks['Prob_Growth'].mean(),
            'Model_Trained_Through': train_end_final,
            'Note': 'OK'
        })

        stock_closes = []
        for ticker in tickers:
            path = os.path.join(stocks_path, f'{ticker.lower()}_clean.csv')
            stock_closes.append(load_price_data(path)['Close'].rename(ticker))

        close_panel = pd.concat(stock_closes, axis=1).reindex(interval_dates).ffill()
        daily_returns = close_panel.pct_change().fillna(0).mean(axis=1)

        for day, daily_ret in daily_returns.items():
            portfolio_value *= (1 + daily_ret)
            portfolio_values.loc[day] = portfolio_value

    spy_plot = spy.loc[start_trading_day:end_trading_day, 'Close']
    spy_plot = spy_plot / spy_plot.iloc[0] * 100
    results = pd.DataFrame({
        'PORTFOLIO': portfolio_values,
        'SP500': spy_plot.reindex(portfolio_values.index).ffill()
    })

    decisions_df = pd.DataFrame(decisions)

    fig, axes = plt.subplots(2, 1, figsize=(14, 9), sharex=False, gridspec_kw={'height_ratios': [3, 1]})

    axes[0].plot(results.index, results['PORTFOLIO'], color='green', lw=3, label='Rotating portfolio')
    axes[0].plot(results.index, results['SP500'], color='black', ls='--', lw=2, label='SP500')
    for rebalance_day in rebalance_dates:
        axes[0].axvline(rebalance_day, color='gray', alpha=0.25, lw=1)
    axes[0].set_title(f'Final strategy from {start_trading_day.date()} to {end_trading_day.date()}')
    axes[0].set_ylabel('Value, start = 100')
    axes[0].grid(True, alpha=0.2)
    axes[0].legend()

    if not decisions_df.empty:
        axes[1].plot(decisions_df['Rebalance_Date'], decisions_df['Mean_Prob_Growth'], marker='o', color='tab:blue')
        axes[1].set_ylim(0, 1)
        axes[1].set_ylabel('Mean Prob_Growth')
        axes[1].set_xlabel('Rebalance date')
        axes[1].grid(True, alpha=0.2)
        for _, row in decisions_df.iterrows():
            axes[1].annotate(row['Selected_Tickers'], (row['Rebalance_Date'], row['Mean_Prob_Growth']),
                             textcoords='offset points', xytext=(0, 8), ha='center', fontsize=8)

    plt.tight_layout()
    plt.show()

    print('\nDECYZJE REBALANCINGU')
    print(f'Retrain each rebalance: {retrain_each_rebalance}')
    print(decisions_df.to_string(index=False))
    print('\nWYNIK')
    print(f"Portfolio Return: {results['PORTFOLIO'].iloc[-1] - 100:.2f}%")
    print(f"SP500 Return:     {results['SP500'].iloc[-1] - 100:.2f}%")
    print(f"Alpha:            {(results['PORTFOLIO'].iloc[-1] - results['SP500'].iloc[-1]):.2f} pp")

    return {
        'model': model,
        'X_train': X_train_final,
        'y_train': y_train_final,
        'results': results,
        'decisions': decisions_df,
        'last_picks': last_picks,
        'last_feature_snapshot': last_feature_snapshot,
        'rebalance_dates': rebalance_dates
    }

final_strategy = run_final_rotating_strategy()
```

## LIME / SHAP dla finalnej wersji

Ta komórka tłumaczy finalny model LightGBM. SHAP pokazuje globalnie, które cechy najmocniej wpływały na decyzje modelu, a LIME pokazuje lokalne uzasadnienie dla konkretnej spółki wybranej przy ostatnim rebalancingu.


```python
# ================================================================
# SHAP + LIME ANALYSIS DLA FINALNEGO MODELU
# ================================================================
model = final_strategy['model']
X_train_final = final_strategy['X_train']
last_picks = final_strategy['last_picks']
last_feature_snapshot = final_strategy['last_feature_snapshot']

# Bierzemy próbkę, żeby wykres był szybki i czytelny.
X_explain = X_train_final.tail(min(1000, len(X_train_final)))

try:
    import shap

    shap_explainer = shap.TreeExplainer(model)
    shap_values = shap_explainer.shap_values(X_explain)

    # Dla klasy binarnej część wersji SHAP zwraca listę [class_0, class_1].
    shap_values_to_plot = shap_values[1] if isinstance(shap_values, list) else shap_values

    shap.summary_plot(shap_values_to_plot, X_explain, plot_type='bar', show=True)
    shap.summary_plot(shap_values_to_plot, X_explain, show=True)
except Exception as exc:
    print(f'SHAP nie uruchomił się w tym środowisku: {exc}')

try:
    from lime.lime_tabular import LimeTabularExplainer

    lime_explainer = LimeTabularExplainer(
        training_data=X_train_final.values,
        feature_names=final_features,
        class_names=['Down / flat', 'Up'],
        mode='classification',
        discretize_continuous=True,
        random_state=42
    )

    if last_picks.empty:
        print('LIME: brak ostatnich wybranych spółek do wyjaśnienia.')
    else:
        ticker_to_explain = last_picks.iloc[0]['Ticker']
        instance = last_feature_snapshot.loc[ticker_to_explain, final_features].values

        def lime_predict_proba(x):
            prob_up = model.predict(pd.DataFrame(x, columns=final_features))
            return np.column_stack([1 - prob_up, prob_up])

        lime_exp = lime_explainer.explain_instance(
            data_row=instance,
            predict_fn=lime_predict_proba,
            num_features=len(final_features)
        )

        print(f'LIME explanation for {ticker_to_explain}:')
        for feature_rule, weight in lime_exp.as_list():
            print(f'{feature_rule}: {weight:.4f}')
except ModuleNotFoundError:
    print('Pakiet lime nie jest zainstalowany. Uruchom w osobnej komórce: !pip install lime')
except Exception as exc:
    print(f'LIME nie uruchomił się w tym środowisku: {exc}')
```

## Jak rozumieć tę logikę?

Tak: model jest trenowany na SPY, czyli uczy się ogólnego zachowania rynku/S&P 500: jakie cechy dzisiaj zwiększały historycznie szansę, że kolejny dzień będzie wzrostowy. Potem ten sam model jest przykładany do sektorów i spółek przez ich aktualne cechy.

W finalnej wersji kod **nie wybiera spółek tylko dlatego, że wyglądają obiecująco po samym trendzie**. Dla każdej spółki przed wyborem liczy cechy i predykcję `Prob_Growth`, czyli szacowaną szansę wzrostu według modelu. Dopiero potem filtruje/waży to kontekstem sektorowym: preferuje spółki z sektorów, które też mają dodatni momentum-score i są nad MA200.

Czyli proces jest dwustopniowy:

1. **Top-down:** najpierw sprawdzamy, które sektory są w lepszej kondycji.
2. **Bottom-up:** w tych sektorach wybieramy spółki, którym model daje najwyższe prawdopodobieństwo wzrostu.

Ważne ograniczenie: `Prob_Growth` jest predykcją krótkoterminową, zasadniczo na kolejny dzień/sygnał, a nie gwarancją dobrego wyniku przez 20 dni. Dlatego rebalance co około 20 dni kalendarzowych jest kompromisem: trzymamy wybór mniej więcej tyle, ile założyłaś, że model ma jeszcze sens, i potem wybieramy od nowa.


### Ważna aktualizacja walk-forward

W poprawionej finalnej wersji przy każdej rotacji model jest domyślnie trenowany od nowa (`retrain_model_each_rebalance = True`). Dla decyzji z dnia `T` używa cech znanych do dnia `T`, ale do treningu bierze tylko wiersze, których target był już znany przed decyzją. To chroni przed look-ahead bias: model nie uczy się na wyniku z przyszłości względem dnia wyboru.

Jeśli drugi miesiąc wypada gorzej niż SP500, to nie musi znaczyć, że kod źle liczy wynik. Może znaczyć, że po pierwszym dobrym oknie zmienił się reżim rynku, SP500 było napędzane przez inne spółki niż wybrane Top 3, filtr sektorowy spóźnił się względem zmiany trendu albo model trenowany na SPY słabo generalizował się na konkretne akcje w tym okresie.


## FINAL v2: model panelowy spółki × daty, target = przyszła 20-dniowa alpha vs SP500

Ta wersja naprawia najważniejszy problem poprzedniej strategii: model nie przewiduje już jednodniowego kierunku SPY, a potem nie próbuje używać tego do miesięcznego wyboru spółek.

Nowa logika:

- budujemy panel `wszystkie_spółki × wszystkie_daty`,
- cechy zawierają: cechy spółki, cechy jej sektora i cechy SPY,
- target to `Future_20d_Alpha = przyszły 20-dniowy zwrot spółki - przyszły 20-dniowy zwrot SPY`,
- model LightGBM uczy się bezpośrednio wybierać spółki, które mają szansę pobić SP500 w horyzoncie ok. 20 sesji,
- przy każdym rebalancingu model jest trenowany walk-forward tylko na etykietach, które byłyby już znane,
- jeśli model nie widzi dodatniej przewagi, portfel przechodzi w SPY,
- zamiast Top 3 używamy bardziej stabilnego Top 5 z wagami odwrotnie proporcjonalnymi do zmienności.


```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import lightgbm as lgb

warnings.filterwarnings("ignore")

# ================================================================
# FINAL v2 CONFIG
# ================================================================
etf_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/ETF clean'
stocks_path = '/Users/marysia/Desktop/studia2/ml_project/Data_clean/Stocks clean'
spy_path = os.path.join(etf_path, 'spy_clean.csv')

target_date = '2015-03-01'
forecast_calendar_days = 60
rebalance_calendar_days = 20
prediction_horizon_trading_days = 20

top_n = 5
max_per_sector = 2
min_predicted_alpha = 0.0000       # 0.00 = wybieramy tylko dodatnią oczekiwaną alfę
min_sector_score = -0.10           # miękki filtr; sektor nie musi być idealny, ale nie może być bardzo słaby
use_inverse_vol_weights = True

sector_map = {
    'AAPL': 'XLK', 'MSFT': 'XLK', 'NVDA': 'XLK', 'AVGO': 'XLK', 'ORCL': 'XLK', 'ADBE': 'XLK',
    'LLY': 'XLV', 'JNJ': 'XLV', 'UNH': 'XLV', 'ABBV': 'XLV', 'MRK': 'XLV',
    'JPM': 'XLF', 'BAC': 'XLF', 'WFC': 'XLF', 'GS': 'XLF', 'MS': 'XLF',
    'GOOGL': 'XLC', 'META': 'XLC', 'NFLX': 'XLC', 'DIS': 'XLC',
    'AMZN': 'XLY', 'TSLA': 'XLY', 'HD': 'XLY', 'NKE': 'XLY',
    'PG': 'XLP', 'KO': 'XLP', 'PEP': 'XLP'
}

panel_features = [
    # stock features
    'Stock_Return_1d', 'Stock_Momentum_5d', 'Stock_Momentum_20d', 'Stock_Momentum_60d',
    'Stock_Volatility_20d', 'Stock_Dist_MA50', 'Stock_Dist_MA200', 'Stock_Vol_Change',
    'Stock_Intraday_Range', 'Stock_Relative_Strength_20d',
    # sector features
    'Sector_Return_1d', 'Sector_Momentum_20d', 'Sector_Volatility_20d', 'Sector_Dist_MA200',
    'Sector_Score_20d',
    # market features
    'SPY_Return_1d', 'SPY_Momentum_20d', 'SPY_Volatility_20d', 'SPY_Dist_MA200'
]


def load_price_data(path):
    df = pd.read_csv(path, index_col='Date', parse_dates=True).sort_index()
    df = df[~df.index.duplicated(keep='last')]
    return df


def add_market_features(df, prefix):
    out = df.copy()
    out[f'{prefix}_Return_1d'] = np.log(out['Close'] / out['Close'].shift(1))
    out[f'{prefix}_Momentum_5d'] = np.log(out['Close'] / out['Close'].shift(5))
    out[f'{prefix}_Momentum_20d'] = np.log(out['Close'] / out['Close'].shift(20))
    out[f'{prefix}_Momentum_60d'] = np.log(out['Close'] / out['Close'].shift(60))
    out[f'{prefix}_Volatility_20d'] = out[f'{prefix}_Return_1d'].rolling(20).std()
    out[f'{prefix}_MA50'] = out['Close'].rolling(50).mean()
    out[f'{prefix}_MA200'] = out['Close'].rolling(200).mean()
    out[f'{prefix}_Dist_MA50'] = (out['Close'] - out[f'{prefix}_MA50']) / out[f'{prefix}_MA50']
    out[f'{prefix}_Dist_MA200'] = (out['Close'] - out[f'{prefix}_MA200']) / out[f'{prefix}_MA200']
    if 'Volume' in out.columns:
        out[f'{prefix}_Vol_Change'] = out['Volume'].pct_change().replace([np.inf, -np.inf], np.nan)
    if {'High', 'Low', 'Open'}.issubset(out.columns):
        out[f'{prefix}_Intraday_Range'] = (out['High'] - out['Low']) / out['Open']
    return out


def build_sector_feature_store():
    sector_store = {}
    for sector in sorted(set(sector_map.values())):
        path = os.path.join(etf_path, f'{sector.lower()}_clean.csv')
        if not os.path.exists(path):
            continue
        df = add_market_features(load_price_data(path), 'Sector')
        df['Sector_Score_20d'] = df['Sector_Momentum_20d'] / (df['Sector_Volatility_20d'] * np.sqrt(20))
        sector_cols = [
            'Sector_Return_1d', 'Sector_Momentum_20d', 'Sector_Volatility_20d',
            'Sector_Dist_MA200', 'Sector_Score_20d'
        ]
        sector_store[sector] = df[sector_cols]
    return sector_store


def build_panel_dataset(horizon=prediction_horizon_trading_days):
    spy = add_market_features(load_price_data(spy_path), 'SPY')
    spy_future_20d = np.log(spy['Close'].shift(-horizon) / spy['Close']).rename('SPY_Future_20d_Return')
    spy_features = spy[['SPY_Return_1d', 'SPY_Momentum_20d', 'SPY_Volatility_20d', 'SPY_Dist_MA200']].copy()
    sector_store = build_sector_feature_store()

    panel_parts = []
    for ticker, sector in sector_map.items():
        path = os.path.join(stocks_path, f'{ticker.lower()}_clean.csv')
        if not os.path.exists(path) or sector not in sector_store:
            continue

        stock = add_market_features(load_price_data(path), 'Stock')
        stock['Future_20d_Return'] = np.log(stock['Close'].shift(-horizon) / stock['Close'])
        stock['Ticker'] = ticker
        stock['Sector'] = sector

        needed_stock_cols = [
            'Ticker', 'Sector', 'Close', 'Future_20d_Return',
            'Stock_Return_1d', 'Stock_Momentum_5d', 'Stock_Momentum_20d', 'Stock_Momentum_60d',
            'Stock_Volatility_20d', 'Stock_Dist_MA50', 'Stock_Dist_MA200', 'Stock_Vol_Change',
            'Stock_Intraday_Range'
        ]
        part = stock[needed_stock_cols].join(sector_store[sector], how='left').join(spy_features, how='left').join(spy_future_20d, how='left')
        part['Future_20d_Alpha'] = part['Future_20d_Return'] - part['SPY_Future_20d_Return']
        part['Stock_Relative_Strength_20d'] = part['Stock_Momentum_20d'] - part['SPY_Momentum_20d']
        panel_parts.append(part)

    panel = pd.concat(panel_parts).sort_index()
    panel.index.name = 'Date'
    panel = panel.replace([np.inf, -np.inf], np.nan)
    return panel


def nearest_trading_day(index, date_like):
    date_like = pd.Timestamp(date_like)
    valid = index[index >= date_like]
    if len(valid) == 0:
        raise ValueError(f'Brak danych giełdowych po dacie {date_like.date()}')
    return valid[0]


def train_panel_alpha_model(panel, as_of_date, horizon=prediction_horizon_trading_days):
    trading_dates = pd.Index(sorted(panel.index.unique()))
    as_of_date = nearest_trading_day(trading_dates, as_of_date)
    as_of_pos = trading_dates.get_loc(as_of_date)
    if as_of_pos <= horizon:
        raise ValueError('Za mało historii przed datą decyzji.')

    # Bez look-ahead: target dla wiersza D zna przyszłość D+horizon.
    # W dniu decyzji T możemy trenować tylko na D <= T-horizon.
    train_label_cutoff = trading_dates[as_of_pos - horizon]
    train = panel.loc[:train_label_cutoff].dropna(subset=panel_features + ['Future_20d_Alpha'])
    if len(train) < 1000:
        raise ValueError(f'Za mało danych panelowych do treningu przed {as_of_date.date()}. Rows: {len(train)}')

    X_train = train[panel_features]
    y_train = train['Future_20d_Alpha']

    params = {
        'objective': 'regression',
        'metric': 'rmse',
        'boosting_type': 'gbdt',
        'learning_rate': 0.03,
        'num_leaves': 21,
        'max_depth': 5,
        'min_data_in_leaf': 40,
        'feature_fraction': 0.85,
        'bagging_fraction': 0.85,
        'bagging_freq': 5,
        'lambda_l1': 0.1,
        'lambda_l2': 0.2,
        'seed': 42,
        'verbose': -1
    }
    model = lgb.train(params, lgb.Dataset(X_train, label=y_train), num_boost_round=350)
    return model, X_train, y_train, train_label_cutoff


def select_panel_portfolio(panel, model, as_of_date):
    rows = panel.loc[[as_of_date]].dropna(subset=panel_features).copy()
    if rows.empty:
        return pd.DataFrame()

    rows['Predicted_20d_Alpha'] = model.predict(rows[panel_features])
    rows['Risk_Adjusted_Score'] = rows['Predicted_20d_Alpha'] / rows['Stock_Volatility_20d'].clip(lower=0.001)

    candidates = rows[
        (rows['Predicted_20d_Alpha'] > min_predicted_alpha) &
        (rows['Sector_Score_20d'] > min_sector_score)
    ].copy()

    if candidates.empty:
        return pd.DataFrame()

    candidates = candidates.sort_values(['Risk_Adjusted_Score', 'Predicted_20d_Alpha'], ascending=False)

    selected_parts = []
    sector_counts = {}
    for _, row in candidates.iterrows():
        sector = row['Sector']
        if sector_counts.get(sector, 0) >= max_per_sector:
            continue
        selected_parts.append(row)
        sector_counts[sector] = sector_counts.get(sector, 0) + 1
        if len(selected_parts) >= top_n:
            break

    if not selected_parts:
        return pd.DataFrame()

    selected = pd.DataFrame(selected_parts)
    selected.index.name = 'Date'

    if use_inverse_vol_weights:
        inv_vol = 1 / selected['Stock_Volatility_20d'].clip(lower=0.001)
        selected['Weight'] = inv_vol / inv_vol.sum()
    else:
        selected['Weight'] = 1 / len(selected)

    return selected


def build_rebalance_dates(trading_index, start_date, end_date, every_calendar_days=20):
    dates = []
    current = pd.Timestamp(start_date)
    end_date = pd.Timestamp(end_date)
    while current <= end_date:
        day = nearest_trading_day(trading_index, current)
        if day <= end_date and (not dates or day != dates[-1]):
            dates.append(day)
        current = current + pd.Timedelta(days=every_calendar_days)
    return dates


def run_panel_alpha_strategy():
    print('Buduję panel danych: wszystkie spółki x wszystkie daty...')
    panel = build_panel_dataset()
    spy = load_price_data(spy_path)

    start_day = nearest_trading_day(spy.index, target_date)
    end_calendar_day = start_day + pd.Timedelta(days=forecast_calendar_days)
    end_day = spy.index[spy.index <= end_calendar_day][-1]
    rebalance_dates = build_rebalance_dates(spy.index, start_day, end_day, rebalance_calendar_days)

    portfolio_values = pd.Series(dtype=float)
    portfolio_value = 100.0
    decisions = []
    last_model = None
    last_X_train = None
    last_y_train = None
    last_selection = pd.DataFrame()

    for i, rebalance_day in enumerate(rebalance_dates):
        next_rebalance = rebalance_dates[i + 1] if i + 1 < len(rebalance_dates) else end_day
        interval_dates = spy.loc[rebalance_day:next_rebalance].index
        if len(interval_dates) == 0:
            continue

        model, X_train, y_train, train_cutoff = train_panel_alpha_model(panel, rebalance_day)
        selected = select_panel_portfolio(panel, model, rebalance_day)
        last_model, last_X_train, last_y_train, last_selection = model, X_train, y_train, selected

        if selected.empty:
            # Fallback: jeśli model nie widzi dodatniej alfy, trzymaj SPY zamiast wymuszać słabe akcje.
            spy_interval = spy.loc[interval_dates, 'Close']
            daily_returns = spy_interval.pct_change().fillna(0)
            tickers_txt = 'SPY fallback'
            sectors_txt = 'SPY'
            mean_alpha = np.nan
            weighted_alpha = np.nan
        else:
            tickers = selected['Ticker'].tolist()
            weights = selected.set_index('Ticker')['Weight']
            close_panel = []
            for ticker in tickers:
                stock_close = load_price_data(os.path.join(stocks_path, f'{ticker.lower()}_clean.csv'))['Close'].rename(ticker)
                close_panel.append(stock_close)
            close_panel = pd.concat(close_panel, axis=1).reindex(interval_dates).ffill()
            daily_returns = close_panel.pct_change().fillna(0).mul(weights, axis=1).sum(axis=1)
            tickers_txt = ', '.join(tickers)
            sectors_txt = ', '.join(sorted(selected['Sector'].unique()))
            mean_alpha = selected['Predicted_20d_Alpha'].mean()
            weighted_alpha = (selected['Predicted_20d_Alpha'] * selected['Weight']).sum()

        for day, ret in daily_returns.items():
            portfolio_value *= (1 + ret)
            portfolio_values.loc[day] = portfolio_value

        decisions.append({
            'Rebalance_Date': rebalance_day,
            'Selected_Tickers': tickers_txt,
            'Selected_Sectors': sectors_txt,
            'Mean_Pred_20d_Alpha': mean_alpha,
            'Weighted_Pred_20d_Alpha': weighted_alpha,
            'Model_Trained_Through': train_cutoff,
            'Rows_Used_For_Training': len(X_train),
            'Note': 'SPY fallback' if selected.empty else 'OK'
        })

    spy_plot = spy.loc[start_day:end_day, 'Close']
    spy_plot = spy_plot / spy_plot.iloc[0] * 100
    results = pd.DataFrame({
        'PORTFOLIO': portfolio_values,
        'SP500': spy_plot.reindex(portfolio_values.index).ffill()
    })
    decisions_df = pd.DataFrame(decisions)

    fig, axes = plt.subplots(2, 1, figsize=(15, 9), sharex=False, gridspec_kw={'height_ratios': [3, 1]})
    axes[0].plot(results.index, results['PORTFOLIO'], color='green', lw=3, label='Panel alpha portfolio')
    axes[0].plot(results.index, results['SP500'], color='black', ls='--', lw=2, label='SP500')
    for day in rebalance_dates:
        axes[0].axvline(day, color='gray', alpha=0.25, lw=1)
    axes[0].set_title(f'Panel alpha strategy: {start_day.date()} to {end_day.date()}')
    axes[0].set_ylabel('Value, start = 100')
    axes[0].grid(True, alpha=0.2)
    axes[0].legend()

    axes[1].plot(decisions_df['Rebalance_Date'], decisions_df['Weighted_Pred_20d_Alpha'], marker='o', color='tab:blue')
    axes[1].axhline(0, color='black', lw=1, alpha=0.4)
    axes[1].set_ylabel('Weighted predicted alpha')
    axes[1].set_xlabel('Rebalance date')
    axes[1].grid(True, alpha=0.2)
    for _, row in decisions_df.iterrows():
        axes[1].annotate(row['Selected_Tickers'], (row['Rebalance_Date'], row['Weighted_Pred_20d_Alpha'] if pd.notna(row['Weighted_Pred_20d_Alpha']) else 0),
                         textcoords='offset points', xytext=(0, 8), ha='center', fontsize=8)

    plt.tight_layout()
    plt.show()

    print('\nDECYZJE REBALANCINGU - PANEL ALPHA MODEL')
    print(decisions_df.to_string(index=False))
    print('\nWYNIK')
    print(f"Portfolio Return: {results['PORTFOLIO'].iloc[-1] - 100:.2f}%")
    print(f"SP500 Return:     {results['SP500'].iloc[-1] - 100:.2f}%")
    print(f"Alpha:            {results['PORTFOLIO'].iloc[-1] - results['SP500'].iloc[-1]:.2f} pp")

    return {
        'panel': panel,
        'model': last_model,
        'X_train': last_X_train,
        'y_train': last_y_train,
        'last_selection': last_selection,
        'decisions': decisions_df,
        'results': results,
        'features': panel_features
    }

panel_alpha_strategy = run_panel_alpha_strategy()
```

## SHAP / LIME dla FINAL v2

Ta komórka wyjaśnia model panelowy, czyli ten, który przewiduje przyszłą 20-dniową alfę spółek względem SP500. SHAP pokazuje globalnie, które cechy były najważniejsze, a LIME lokalnie tłumaczy ostatnio wybraną spółkę.


```python
# ================================================================
# SHAP + LIME DLA FINAL v2 PANEL ALPHA MODEL
# ================================================================
model = panel_alpha_strategy['model']
X_train_panel = panel_alpha_strategy['X_train']
last_selection = panel_alpha_strategy['last_selection']
features_panel = panel_alpha_strategy['features']

X_explain = X_train_panel.tail(min(1500, len(X_train_panel)))

try:
    import shap
    shap_explainer = shap.TreeExplainer(model)
    shap_values = shap_explainer.shap_values(X_explain)
    shap.summary_plot(shap_values, X_explain, plot_type='bar', show=True)
    shap.summary_plot(shap_values, X_explain, show=True)
except Exception as exc:
    print(f'SHAP nie uruchomił się w tym środowisku: {exc}')

try:
    from lime.lime_tabular import LimeTabularExplainer

    if last_selection.empty:
        print('LIME: ostatnia decyzja była SPY fallback, więc nie ma pojedynczej spółki do wyjaśnienia.')
    else:
        ticker_to_explain = last_selection.iloc[0]['Ticker']
        instance = last_selection.iloc[0][features_panel].astype(float).values

        lime_explainer = LimeTabularExplainer(
            training_data=X_train_panel.values,
            feature_names=features_panel,
            mode='regression',
            discretize_continuous=True,
            random_state=42
        )

        def lime_predict(x):
            return model.predict(pd.DataFrame(x, columns=features_panel))

        lime_exp = lime_explainer.explain_instance(
            data_row=instance,
            predict_fn=lime_predict,
            num_features=10
        )

        print(f'LIME explanation for {ticker_to_explain} - predicted 20d alpha:')
        for feature_rule, weight in lime_exp.as_list():
            print(f'{feature_rule}: {weight:.5f}')
except ModuleNotFoundError:
    print('Pakiet lime nie jest zainstalowany. Uruchom w osobnej komórce: !pip install lime')
except Exception as exc:
    print(f'LIME nie uruchomił się w tym środowisku: {exc}')
```

## Co zmieniło się względem poprzedniej wersji?

Poprzednia wersja trenowała model na SPY i przewidywała krótkoterminowe prawdopodobieństwo wzrostu. To było słabo dopasowane do zadania wyboru akcji na około 20 dni.

FINAL v2 trenuje model na wielu spółkach i wielu datach. Target jest dokładnie tym, czego potrzebuje strategia: czy dana spółka pobije SP500 w następnym 20-dniowym oknie. Dzięki temu model nie pyta już „czy coś wzrośnie jutro?”, tylko „czy ta spółka ma szansę dać dodatnią alfę względem indeksu w horyzoncie trzymania?”.

Jeśli wynik nadal będzie słabszy od SP500, kolejne miejsca do sprawdzenia to: próg `min_predicted_alpha`, liczba spółek `top_n`, filtr sektorowy `min_sector_score`, horyzont `prediction_horizon_trading_days` oraz porównanie predykowanej alfy z faktyczną alfą po każdym rebalancingu.
